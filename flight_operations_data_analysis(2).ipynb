```python
# Flight Operations Data Analysis

## Day 10 - Python Programming Internship

This notebook performs a complete data analysis workflow on a Flight Operations dataset using Python and Pandas.

### Objectives

- Load and understand the flight operations dataset
- Inspect its structure and statistics
- Select and filter relevant records
- Sort flight records
- Perform grouping and aggregation
- Apply suitable data transformations
- Identify patterns and key findings
- Write meaningful observations based on the analysis
```

```python
import pandas as pd
import numpy as np
from google.colab import files

# Upload the Flight Operations CSV file
uploaded = files.upload()

# Get the uploaded filename
file_name = next(iter(uploaded))

# Load the dataset
df = pd.read_csv(file_name)

print("Dataset loaded successfully!")
print("File:", file_name)
```

```python
## 1. Dataset Overview

Let's inspect the structure, dimensions, columns, data types, and statistical characteristics of the Flight Operations dataset.
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 Rows:")
display(df.head())

print("\nLast 5 Rows:")
display(df.tail())

print("\nRandom Sample:")
display(df.sample(5))

print("\nDataset Information:")
df.info()

print("\nDescriptive Statistics:")
display(df.describe())
```

```python
## 2. Data Quality Check

Before performing analysis, we check for missing values, duplicate records, and data types.
```

```python
print("Data Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nTotal Missing Values:", df.isnull().sum().sum())

print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
# Convert Flight_Date to DateTime
df["Flight_Date"] = pd.to_datetime(df["Flight_Date"])

print("\nFlight date range:")
print("Start:", df["Flight_Date"].min())
print("End:", df["Flight_Date"].max())
```

```python
## 3. Data Selection, Filtering and Sorting

Pandas can be used to select relevant columns, filter records based on conditions, and sort flights to identify important operational patterns.
```

```python
# Select relevant flight information
selected_data = df[
    [
        "Flight_ID",
        "Airline",
        "Origin",
        "Destination",
        "Passengers",
        "Delay_Minutes",
        "Flight_Status",
        "Passenger_Satisfaction"
    ]
]

display(selected_data.head(10))
```

```python
delayed_flights = df[df["Flight_Status"] == "Delayed"]

print("Number of delayed flights:", len(delayed_flights))
display(delayed_flights.head(10))
```

```python
high_delay_flights = df[df["Delay_Minutes"] >= 30]

print("Flights delayed by 30 minutes or more:", len(high_delay_flights))
display(
    high_delay_flights.sort_values(
        by="Delay_Minutes",
        ascending=False
    ).head(10)
)
```

```python
print("Top flights by passenger satisfaction:")
display(
    df.sort_values(
        by="Passenger_Satisfaction",
        ascending=False
    ).head(10)
)
```

```python
## 4. Grouping and Aggregation

Grouping allows us to compare flight operations across airlines, travel classes, flight statuses, and other categories.
```

```python
airline_analysis = df.groupby("Airline").agg(
    Total_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Average_Passengers=("Passengers", "mean"),
    Average_Delay=("Delay_Minutes", "mean"),
    Average_Ticket_Price=("Average_Ticket_Price", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
).sort_values(
    by="Total_Passengers",
    ascending=False
)

display(airline_analysis)
```

```python
origin_analysis = df.groupby("Origin").agg(
    Total_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Average_Delay=("Delay_Minutes", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
).sort_values(
    by="Total_Flights",
    ascending=False
)

display(origin_analysis)
```

```python
destination_analysis = df.groupby("Destination").agg(
    Total_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Average_Delay=("Delay_Minutes", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
).sort_values(
    by="Total_Passengers",
    ascending=False
)

display(destination_analysis)
```

```python
## 5. Operational Pattern Analysis

Let's examine flight status, travel class, and weather conditions to identify operational patterns.
```

```python
status_analysis = df.groupby("Flight_Status").agg(
    Number_of_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Average_Delay=("Delay_Minutes", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
).sort_values(
    by="Number_of_Flights",
    ascending=False
)

display(status_analysis)
```

```python
class_analysis = df.groupby("Travel_Class").agg(
    Number_of_Flights=("Flight_ID", "count"),
    Total_Passengers=("Passengers", "sum"),
    Average_Ticket_Price=("Average_Ticket_Price", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
).sort_values(
    by="Total_Passengers",
    ascending=False
)

display(class_analysis)
```

```python
weather_analysis = df.groupby("Weather").agg(
    Number_of_Flights=("Flight_ID", "count"),
    Average_Delay=("Delay_Minutes", "mean"),
    Average_Satisfaction=("Passenger_Satisfaction", "mean")
).sort_values(
    by="Average_Delay",
    ascending=False
)

display(weather_analysis)
```

```python
## 6. Key Flight Operations Statistics

Let's answer several meaningful questions using the dataset.
```

```python
# Overall statistics
total_flights = df["Flight_ID"].nunique()
total_passengers = df["Passengers"].sum()
average_passengers = df["Passengers"].mean()
average_delay = df["Delay_Minutes"].mean()
maximum_delay = df["Delay_Minutes"].max()
average_ticket_price = df["Average_Ticket_Price"].mean()
average_satisfaction = df["Passenger_Satisfaction"].mean()

print(f"Total Flights: {total_flights}")
print(f"Total Passengers: {total_passengers:,}")
print(f"Average Passengers per Flight: {average_passengers:.2f}")
print(f"Average Delay: {average_delay:.2f} minutes")
print(f"Maximum Delay: {maximum_delay:.2f} minutes")
print(f"Average Ticket Price: ₹{average_ticket_price:.2f}")
print(f"Average Passenger Satisfaction: {average_satisfaction:.2f}/5")
```

```python
most_delayed_airline = (
    df.groupby("Airline")["Delay_Minutes"]
    .mean()
    .sort_values(ascending=False)
)

print("Airlines ranked by average delay:")
display(most_delayed_airline)
```

```python
highest_passenger_airline = (
    df.groupby("Airline")["Passengers"]
    .sum()
    .sort_values(ascending=False)
)

print("Airlines ranked by total passengers:")
display(highest_passenger_airline)
```

```python
most_common_status = df["Flight_Status"].value_counts()

print("Flight Status Distribution:")
display(most_common_status)
```

```python
booking_analysis = df["Booking_Channel"].value_counts()

print("Booking Channel Distribution:")
display(booking_analysis)
```

```python
weather_delay = (
    df.groupby("Weather")["Delay_Minutes"]
    .mean()
    .sort_values(ascending=False)
)

print("Average delay by weather condition:")
display(weather_delay)
```
